```python
# Cell 1
#Class management System
Class_attendance_list={'Student1':12, 'Student2':15, 'Student3':14, 'Student4':10, 'Student5':11}
Total_classes=15
```

```python
# Cell 2
while True:
    print()
    print('''What do you want to do?
1) Mark attendance
2) Add a student to class
3) Quit''')
    n=input('Enter your choice: ')
    if not n.isdigit():
        print('Please enter a number between 1 and 3')
        continue
    n=int(n)
    if n==1:
        Total_classes+=1
        print('Attendance for class number', Total_classes)
        for name in Class_attendance_list:
            a=input(name+' present? (yes/no): ').lower()
            while a!='yes' and a!='no':
                a=input('Please type yes or no: ').lower()
            if a=='yes':
                Class_attendance_list[name]+=1

        print('Attendance so far')
        for name in Class_attendance_list:
            att=Class_attendance_list[name]
            per=round(att/Total_classes*100, 1)
            low=''
            if per<75:
                print('Low attendance')
    elif n==2:
        name=input('Enter the name of the student: ').strip()
        if name=='':
            print('Name cannot be empty')
        elif name in Class_attendance_list:
            print(name, 'is already in the class')
        else:
            att=input('How many classes has '+name+' attended till now? ')
            if att.isdigit() and int(att)<=Total_classes:
                Class_attendance_list[name]=int(att)
                print(name, 'added to the class')
            else:
                print('Please enter a number between 0 and', Total_classes)
    elif n==3:
        print('Thank you, bye!')
        break

    else:
        print('Please enter a number between 1 and 3')
```
